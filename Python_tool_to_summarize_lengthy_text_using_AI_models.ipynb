```python
# Import the Summarization Pipeline
from transformers import pipeline
```

```python
# Create the Summarizer
summarizer = pipeline("summarization")
```

```python
#. Provide Long Text to Summarize
long_text = """ Artificial Intelligence (AI) is a transformative technology that enables machines to simulate human intelligence, encompassing the abilities to learn, reason, and solve problems. It operates through algorithms and data-driven models that allow computers to perform complex tasks such as recognizing speech, analyzing images, translating languages, and making predictions. The history of AI dates back to the mid-20th century when pioneers like John McCarthy and Alan Turing envisioned machines capable of thinking, eventually leading to breakthroughs like IBM’s Deep Blue defeating chess champion Garry Kasparov and Google’s AlphaGo defeating a Go master. Today, AI is deeply integrated into daily life and industries—healthcare systems use it to detect diseases early; financial institutions rely on it for fraud detection and predictive analytics; autonomous vehicles navigate safely through city traffic; and virtual assistants like Siri or Alexa enhance user convenience through voice interaction. At the heart of AI lies machine learning and deep learning, which utilize data to train models that adapt and improve their accuracy over time, enhancing decision-making across disciplines. The applications are vast: in education, AI personalizes learning; in manufacturing, it ensures quality control; in agriculture, it optimizes crop yield; and in entertainment, it powers recommendations and content creation. The major benefits include increased efficiency, precision, innovation, and accessibility, making progress faster and more inclusive. However, AI also presents challenges such as job displacement due to automation, algorithmic bias from skewed data, threats to data privacy, and a lack of transparency in decision-making processes. As AI systems become more powerful, ethical considerations grow increasingly important—ensuring fairness, accountability, and responsible use. The future of AI holds immense promise, with advancements likely to revolutionize sectors such as medicine, climate science, and creative industries. Generative AI, for instance, now creates text, music, and art, expanding human creativity beyond traditional limits. But to fully realize its potential, society must govern AI with ethical frameworks that protect human values and promote trust. In essence, AI is not a replacement for human intelligence but an extension of it—when used wisely, it amplifies creativity, accelerates progress, and builds a smarter and more sustainable world for everyone.
"""
```

```python
# Cheak the length of the content
(len(long_text))
```

```python
summary = summarizer(long_text, max_length=250, min_length=50, do_sample=False, truncation=True)
print("Summary:", summary[0]['summary_text'])
```
